## 🛡️ 3. Il Player Didattico Anti-Skip ad Alta Sicurezza
---
Il visualizzatore video (sviluppato in Alpine.js) include **4 livelli di protezione attiva**:

1. **Blocco Seekbar (Anti-Skip):** Impossibile trascinare il cursore in avanti. È consentito solo il riavvolgimento nelle parti già viste.
2. **Pausa Automatica al Cambio Scheda (`strict_tab_pause`):** Se il corsista cambia tab o minimizza il browser, il video e il tempo si arrestano all'istante.
3. **Pop-up di Verifica Presenza Attiva:** Comparsa di finestre interattive a intervalli con conto alla rovescia (60 secondi); se l'utente non conferma, il video si blocca.
4. **YouTube Shield No-Logo:** Possibilità di utilizzare file MP4 o video YouTube senza loghi, canali o link esterni.

````python
# Simulazione interattiva della logica di calcolo frequenza soglia 80%
def verifica_idoneita(nome, ore_zoom, ore_fad, ore_totali_corso=25, soglia_min=80.0):
    ore_totali_frequentate = ore_zoom + ore_fad
    percentuale = (ore_totali_frequentate / ore_totali_corso) * 100
    idoneo = percentuale >= soglia_min
    stato = "🟢 IDONEO (Attestato Sbloccato)" if idoneo else "🔴 NON IDONEO (Sotto soglia 80%)"
    return {
        "Corsista": nome,
        "Zoom (h)": ore_zoom,
        "FAD (h)": ore_fad,
        "Totale (h)": ore_totali_frequentate,
        "%": f"{percentuale:.1f}%",
        "Esito Ministeriale": stato
    }

studenti_esempio = [
    verifica_idoneita("Elena Bianchi", 20.0, 5.0),
    verifica_idoneita("Marco Valeri", 19.0, 3.5),
    verifica_idoneita("Sara Lombardi", 17.0, 2.0)
]

import pandas as pd
df = pd.DataFrame(studenti_esempio)
df
````
